# BFS Traversal

BFS is a graph traversal algorithm that:

Visits nodes level by level

Uses a queue

Is great for:

Shortest path in unweighted graphs

Level-order traversal

Finding connected components


Graph Representation (Adjacency List)

We’ll use a dictionary where:

Key → node

Value → list of neighbors
graph = {
    0: [1, 2],
    1: [0, 3, 4],
    2: [0, 5],
    3: [1],
    4: [1],
    5: [2]
}


```python
from collections import deque

from collections import deque

def bfs(graph, start):
    visited = set()
    queue = deque([start])

    visited.add(start)

    while queue:
        node = queue.popleft()
        print(node, end=" ")

        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)

# -------- Driver Code --------
if __name__ == "__main__":
    graph = {
        0: [1, 2],
        1: [0, 3, 4],
        2: [0, 5],
        3: [1],
        4: [1, 5],
        5: [2, 4]
    }

    start_node = 0
    print("BFS Traversal:")
    bfs(graph, start_node)
```

## DFS Traversal 

Depth First Search (DFS) is a graph traversal technique where:

You go as deep as possible along one path before coming back and trying another path.

##### Think of it like exploring a maze:

You pick a path

Keep walking forward

If you hit a dead end → you backtrack

Then try a different path


#### 🧠 Core Idea

DFS follows this simple rule:

Start from a node

Mark it as visited

Visit one unvisited neighbor

Repeat step 2–3 for that neighbor

When no unvisited neighbors remain → backtrack

```python
def dfs(graph, start):
    visited = set()
    stack = [start]

    visited.add(start)

    while stack:
        node = stack.pop()
        print(node, end=" ")

        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                stack.append(neighbor)

graph = {
    0: [1, 2],
    1: [3, 4],
    2: [5],
    3: [],
    4: [],
    5: []
}


start_node = 0
print("DFS Traversal:")
dfs(graph, start_node)
```

### Number of Provinces
🔗 Problem Idea (in simple words)

You are given a matrix isConnected where:

isConnected[i][j] == 1 → city i is directly connected to city j

A province is a group of cities that are directly or indirectly connected

Your task:

Count how many such provinces exist.

isConnected = [
  [1,1,0],
  [1,1,0],
  [0,0,1]
]


```python
class Solution:
    def findCircleNum(self, isConnected):
        n = len(isConnected)
        visited = set()

        def dfs(city):
            for neighbor in range(n):
                if isConnected[city][neighbor] == 1 and neighbor not in visited:
                    visited.add(neighbor)
                    dfs(neighbor)

        provinces = 0

        for city in range(n):
            if city not in visited:
                provinces += 1
                visited.add(city)
                dfs(city)

        return provinces


# -------- Driver Code --------
if __name__ == "__main__":
    solution = Solution()

    isConnected = [
        [1, 1, 0],
        [1, 1, 0],
        [0, 0, 1]
    ]

    result = solution.findCircleNum(isConnected)
    print("Number of Provinces:", result)
```
