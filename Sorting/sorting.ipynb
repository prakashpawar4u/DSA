```python
## Bubble Sort
## Bubbles the biggest element towards the right

from typing import List
def BubbleSort(nums: List[int]) -> bool:
    n = len(nums)
    for i in range(n-1):
        # print(i, nums[i])
        swapped = False
        for j in range(n-i -1):
            # print(f" J values: {j}")
            if nums[j] > nums[j+1]:
                nums[j], nums[j+1] = nums[j+1], nums[j]
                swapped = True
        if swapped == False:
            print(nums)
            return


BubbleSort([4,2,6,1,3,5])
```

```python
## Insertion Sort



from typing import List
def Insertion(nums: List[int]) -> bool:
    n = len(nums)
    for i in range(1, n):
        x = nums[i]
        j = i-1

        while j >= 0 and x < nums[j]:
            nums[j+1] = nums[j]
            j = j -1
        nums[j+1] = x
    print(nums)   

Insertion([4,2,6,1,3,5])
```
