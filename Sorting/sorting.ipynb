```python
## Bubble Sort
## Bubbles the biggest element towards the right

from typing import List
def BubbleSort(nums: List[int]) -> bool:
    n = len(nums)
    for i in range(n-1):
        # print(i, nums[i])
        swapped = False
        for j in range(n-i -1):
            # print(f" J values: {j}")
            if nums[j] > nums[j+1]:
                nums[j], nums[j+1] = nums[j+1], nums[j]
                swapped = True
        if swapped == False:
            print(nums)
            return


BubbleSort([4,2,6,1,3,5])
```

# Insertion Sort

Start from the second element (the first is already “sorted”).

Compare the current element with the one before it.

Shift larger elements one position to the right.

Insert the current element in the correct position.

Repeat until the list is sorted.

```python
## Insertion Sort



from typing import List
def Insertion(nums: List[int]) -> bool:
    n = len(nums)
    for i in range(1, n):
        x = nums[i]
        j = i-1

        while j >= 0 and x < nums[j]:
            nums[j+1] = nums[j]
            j = j -1
        nums[j+1] = x
    print(nums)   

Insertion([4,2,6,1,3,5])
```

# Selection Sort 
selects the smallest element from the unsorted part of the array and swaps it with the first element of that unsorted part.

```python
def selectionSort(nums: List[int]) -> List[int]:
    n = len(nums)

    for i in range(n-1):
        min_ind = i

        #find min element in the unsorted part
        for j in range(i + 1, n):
            if nums[j] < nums[min_ind]:
                min_ind = j

        # Swap only if a smaller element is found
        if min_ind != i:
            nums[i], nums[min_ind] = nums[min_ind], nums[i]
    return nums
data = [8, 3, 5, 2]
print(selectionSort(data))

```
