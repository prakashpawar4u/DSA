# Sum of n natural numbers

Input : n = 3
Output : 6


```python
def sum_of_n(n):
    total = 0
    for i in range(n):
        print(i)
        total +=i
    print(total)

sum_of_n(3)
```

```python
sum_of_n(5)
```
