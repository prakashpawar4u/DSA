# 1. Two Sum

```python
class Solution(object):
    def twoSum(self, nums, target):
        """
        :type nums: List[int]
        :type target: int
        :rtype: List[int]
        """
        hashMap = {}
        for i, n in enumerate(nums):
            diff = target - n
            if diff in hashMap:
                return(hashMap[diff], i)
            hashMap[n] = i
        return
nums = [2, 7, 11, 15]
target = 9
sobj = Solution()
print(sobj.twoSum(nums, target))
```

# 2. Best Time to Buy and Sell Stock 

You are given an array prices where prices[i] is the price of a given stock on the ith day.



```python
from typing import List
def maxProfit(prices: List[int]) -> int:
    min_price = float('inf') #set to infinity initially
    max_profit = 0

    for price in prices:
        if price < min_price:
            min_price = price
        else:
            max_profit = max(max_profit, price - min_price)
    return max_profit 
prices = [7,1,5,3,6,4]
print(maxProfit(prices))
```

# 4. 238. Product of Array Except Self

Precompute prefix products (left side)

Precompute suffix products (right side)
🔹 Approach

Create an output array answer initialized with 1

Traverse from left to right:

answer[i] stores product of all elements before i

Traverse from right to left:

Multiply answer[i] by product of all elements after i

Return answer

```python
def productExceptSelf(nums: List[int]) -> List[int]:
    res = [1] * (len(nums))
    prefix = 1

    for i in range(len(nums)):
        res[i] = prefix
        prefix *= nums[i]
    
    postfix = 1

    for i in range(len(nums)-1, -1, -1):
        res[i] *= postfix
        postfix *= nums[i]
    
    return res

nums = [1,2,3,4]
print(productExceptSelf(nums))
```

# 5. 53. Maximum Subarray

Initialize:

current_sum → sum of the best subarray ending at current index

max_sum → overall maximum subarray sum

Traverse the array:

Update current_sum as the maximum of:

current element alone

current element + previous current_sum

Update max_sum if current_sum is larger

Return max_sum

```python
def maxSubArray(nums: List[int]) -> int:
    maxSub = nums[0]
    curSum = 0

    for num in nums:
        if curSum < 0:
            curSum = 0

        curSum += num
        maxSub = max(maxSub, curSum)
    return maxSub
    
nums = [-2,1,-3,4,-1,2,1,-5,4]
maxSubArray(nums)
```

# 6. 152. Maximum Product Subarray

Approach

Initialize:

max_prod = nums[0] → max product so far

min_prod = nums[0] → min product so far

result = nums[0] → global max product

Traverse the array from index 1:

Temporarily store the old max_prod

Update max_prod = max(current number, current × old max, current × old min)

Update min_prod = min(current number, current × old max, current × old min)

Update result = max(result, max_prod)

Return result



```python
## Maximum Subarray

nums = [2,3,-1,4]


def maxProduct(nums):
    n = len(nums)

    prefix = 1
    suffix = 1

    ans = float('-inf')

    for i in range(n):
        prefix *=nums[i]
        suffix *= nums[n -1 -i]

        ans = max(ans, prefix, suffix)

        if prefix == 0:
            prefix = 1
        if suffix == 0:
            suffix = 1
    return ans

print(maxProduct(nums))
```

# 7 153. Find Minimum in Rotated Sorted Array


🔹 Intuition

A sorted array is rotated. Example: [4,5,6,7,0,1,2]

The minimum element is the only element where the previous element is bigger (or simply the pivot point).

Because the array is mostly sorted, we can use binary search instead of linear scan → O(log n).

🔹 Approach

Initialize left = 0, right = len(nums) - 1.

While left < right:

Find mid = (left + right) // 2

Compare nums[mid] with nums[right]:

If nums[mid] > nums[right] → minimum is in right half, set left = mid + 1

Else → minimum is in left half including mid, set right = mid

When loop ends, left == right → the minimum element.

Key Insight:

If mid > right → min is after mid

Else → min is at mid or before


```python
class Solution(object):
    def findMin(self, nums):
        left, right = 0, len(nums) - 1

        while left < right:
            mid = (left + right) // 2
            if nums[mid] > nums[right]:
                left = mid + 1
            else:
                right = mid

        return nums[left]
nums = [4, 5, 6, 7, 0, 1, 2]
sol = Solution()
print(sol.findMin(nums))

```

# 8. Search in Rotated Sorted Array 


🔹 Intuition

We need to find a target in a rotated sorted array in O(log n) time.

The array is mostly sorted, except for the rotation.

Binary search still works if we figure out which half is sorted at each step:

Check if the left half is sorted: nums[left] <= nums[mid]

Otherwise, the right half is sorted.

If the target lies within the sorted half, search there; else search the other half.

🔹 Approach

Initialize left = 0 and right = len(nums) - 1

While left <= right:

Compute mid = (left + right) // 2

If nums[mid] == target, return mid

Check which half is sorted:

Left half sorted (nums[left] <= nums[mid])

If target is in [nums[left], nums[mid]], move right = mid - 1

Else move left = mid + 1

Right half sorted (nums[mid] < nums[right])

If target is in [nums[mid], nums[right]], move left = mid + 1

Else move right = mid - 1

If not found, return -1

```python
class Solution(object):
    def search(self, nums, target):
        left, right = 0, len(nums) - 1

        while left <= right:
            mid = (left + right) // 2
            if nums[mid] == target:
                return mid
            
            # Left Half is sorted
            if nums[left] <= nums[mid]:
                if nums[left] <= target < nums[mid]:
                    right = mid -1
                else: 
                    left = mid + 1
            # Right half sorted
            else: 
                if nums[mid] < target <= nums[right]:
                    left = mid + 1
                else:
                    right = mid -1 
                        


        return -1


if __name__ == "__main__":
    nums = [4,5,6,7,0,1,2]
    target = 0
    sol = Solution()
    print(sol.search(nums, target))
```

# 9 15. 3Sum
Given an integer array nums, return all the triplets [nums[i], nums[j], nums[k]] such that i != j, i != k, and j != k, and nums[i] + nums[j] + nums[k] == 0. 


✅ Strategy

Sort the array

Fix one number nums[i]

Use two pointers:

left = i + 1

right = n - 1

Move pointers based on the sum

Skip duplicates at every step

```python
class Solution:
    def threeSum(self, nums: list[int]) -> list[list[int]]:
        nums.sort()
        res = []
        n = len(nums)

        for i in range(n):
            if i > 0 and nums[i] == nums[i-1]:
                continue

            left, right = i + 1, n - 1

            while left < right:
                total = nums[i] + nums[left] + nums[right]

                if total == 0:
                    res.append([nums[i], nums[left], nums[right]])

                
                    # skip duplicates for left and right 

                    while left < right and nums[left] == nums[left + 1]:
                        left += 1
                    while left < right and nums[right] == nums[right - 1]:
                        right -= 1
                    
                    left += 1
                    right -= 1

                elif total < 0:
                    left += 1
                else: 
                    right  -= 1
        return res

if __name__ == "__main__":
    nums = [-1,0,1,2,-1,-4]

    sol = Solution()
    print(sol.threeSum(nums))
```

# 10. Container with Most Water 

The amount of water between two lines i and j is:

area = min(height[i], height[j]) * (j - i)

Width = distance between pointers

Height = shorter of the two lines

```python
class Solution:
    def maxArea(self, heights: List[int]) -> int:
        res = 0
        l, r = 0, len(heights)-1

        while l < r:
            area = (r -1) * min(heights[l], heights[r])
            res = max(res, area)

            if heights[l] < heights[r]:
                l += 1
            
            else:
                r -=1
        return res 

height = [1,8,6,2,5,4,8,3,7]
sObj = Solution()
print(sObj.maxArea(height))
```

# 11 Sum of Two Integers:

This hints at using bit manipulation:

XOR (^): gives sum without carrying

Example: 1 ^ 1 = 0 (1 + 1 = 0 with carry)

Example: 1 ^ 0 = 1 (1 + 0 = 1)

AND (&) + left shift (<<): gives the carry bits

Example: 1 & 1 = 1 → shift left → 10 (binary 2)

This is the value to add to sum in next iteration

Repeat until carry becomes 0.


### 🧪 Example Walkthrough

Input: a = 2 (10), b = 3 (11)

a ^ b = 10 ^ 11 = 01 → sum_without_carry = 1

(a & b) << 1 = (10 & 11) << 1 = 10 << 1 = 100 → carry = 4

Next iteration: a = 1, b = 4

1 ^ 4 = 5 → sum_without_carry = 5

(1 & 4) << 1 = 0 << 1 = 0 → carry = 0

Carry = 0 → stop → Answer = 5



```python
class Solution:
    def getSum(self, a: int, b: int) -> int:
        # 32-bit mask for handling negative numbers
        MASK = 0xFFFFFFFF
        MAX_INT = 0x7FFFFFFF

        while b != 0:
            # sum without carry
            sum_ = (a ^ b) & MASK
            # carry
            b = ((a & b) << 1) & MASK
            a = sum_

        # handle negative numbers
        return a if a <= MAX_INT else ~(a ^ MASK)
        

# Example usage
sol = Solution()
print(sol.getSum(1, 2))  # Output: 3
print(sol.getSum(2, 3))  # Output: 5
```

```python
class Solution:
    def countBits(self, n: int):
        ans = [0] * (n + 1)

        for i in range(1, n + 1):
            ans[i] = ans[i >> 1] + (i & 1)

        return ans

sobj = Solution()
print(sobj.countBits(5))
```

Reverse bits of a given 32 bits signed integer.

Example 1:
Input: n = 43261596
Output: 964176192

Explanation:

Integer	Binary

43261596	00000010100101000001111010011100

964176192	00111001011110000010100101000000


```python
def reverseBits(n: int) -> int:
    result = 0 
    for _ in range(32):
        result = (result << 1) | (n & 1)
        n >>= 1
    
    return result

print(reverseBits(43261596))
```

# 16 Climbing Stairs

Example 2:

Input: n = 3
Output: 3
Explanation: There are three ways to climb to the top.
1. 1 step + 1 step + 1 step
2. 1 step + 2 steps
3. 2 steps + 1 step


```python
def climbStairs(n: int) -> int:
    if n == 1:
        return 1
    prev2, prev1 = 1, 1

    for i in range(2, n + 1):
        current = prev1 + prev2 
        prev2 = prev1
        prev1= current
    return prev1

print(climbStairs(3))
```

# 17 Coin Change 

You are given an integer array coins representing coins of different denominations and an integer amount representing a total amount of money.

Return the fewest number of coins that you need to make up that amount. If that amount of money cannot be made up by any combination of the coins, return -1.

You may assume that you have an infinite number of each kind of coin.

 

Example 1:

Input: coins = [1,2,5], amount = 11
Output: 3
Explanation: 11 = 5 + 5 + 1

```python
class Solution:
    def coinChange(self, coins: List[int], amount: int) -> int:
        dp = [float('inf')] * (amount + 1)
        dp[0] = 0  # base case

        for a in range(1, amount + 1):
            for coin in coins:
                if a - coin >= 0:
                    dp[a] = min(dp[a], 1 + dp[a - coin])

        return dp[amount] if dp[amount] != float('inf') else -1
    
sobj = Solution()
coins = [1,2,5]
amount = 11
print(sobj.coinChange(coins, amount))
```

# 18 300. Longest Increasing Subsequence

Given an integer array nums, return the length of the longest strictly increasing subsequence.

 

Example 1:

Input: nums = [10,9,2,5,3,7,101,18]
Output: 4

Explanation: The longest increasing subsequence is [2,3,7,101], therefore the length is 4.
Example 2:

Input: nums = [0,1,0,3,2,3]
Output: 4
Example 3:

Input: nums = [7,7,7,7,7,7,7]
Output: 1

```python
class Solution:
    def lengthOfLIS(self, nums: List[int]) -> int:
        n = len(nums)
        dp = [1] * n

        for i in range(n):
            for j in range(i):
                if nums[j] < nums[i]:
                    dp[i] = max(dp[i], dp[j] + 1)
        return max(dp)


nums = [10,9,2,5,3,7,101,18]
sobj = Solution()
print(sobj.lengthOfLIS(nums))
```

```python
# Optimized Code 

from bisect import bisect_left 
class Solution:
    def lengthOfLIS(self, nums: List[int]) -> int:
        tails = []

        for num in nums:
            idx = bisect_left(tails, num)
            if idx == len(tails):
                tails.append(num)
            else:
                tails[idx] = num
        return len(tails)
    
nums = [10,9,2,5,3,7,101,18]
sobj = Solution()
print(sobj.lengthOfLIS(nums))
```

# 19 1143 Longest Common Subsequence

Given two strings text1 and text2, return the length of their longest common subsequence. If there is no common subsequence, return 0
Example 1:

Input: text1 = "abcde", text2 = "ace" 
Output: 3  
Explanation: The longest common subsequence is "ace" and its length is 3.


```python
class Solution:
    def longestCommonSubsequence(self, text1: str, text2: str) -> int:
        m = len(text1)
        n = len(text2)

        dp = [[0] * (n+1) for _ in range(m+1)]

        for i in range(1, m + 1):
            for j in range(1, n + 1):
                if text1[i - 1] == text2[j - 1]:
                    dp[i][j] = 1 + dp[i - 1][j - 1]
                else:
                    dp[i][j] = max(dp[i-1][j], dp[i][j - 1])
        
        return dp[m][n]
    

    
text1 = "abcde"
text2 = "ace" 
sobj = Solution()
print(sobj.longestCommonSubsequence(text1, text2))
```

# 20  139. Word Break
Given a string s and a dictionary of strings wordDict, return true if s can be segmented into a space-separated sequence of one or more dictionary words.

Note that the same word in the dictionary may be reused multiple times in the segmentation.

 

Example 1:

Input: s = "leetcode", wordDict = ["leet","code"]
Output: true
Explanation: Return true because "leetcode" can be segmented as "leet code".

```python
class Solution:
    def wordBreak(self, s: str, wordDict: List[str]) -> bool:
        n = len(s)
        word_set = set(wordDict)

        dp = [False] * (n+1)
        dp[n] = True

        for i in range(n -1, -1, -1):
            for word in word_set:
                length = len(word)
                if i + length <= n and s[i: i+length] == word:
                    if dp[i + length]:
                        dp[i] = True
                        break
        return dp[0]

sobj = Solution()
sobj.wordBreak(s = "leetcode", wordDict = ["leet","code"])
    
```

```python
class Solution:
    def wordBreak(self, s: str, wordDict: List[str]) -> bool:
        word_set = set(wordDict)     # O(1) lookup
        n = len(s)

        dp = [False] * (n + 1)
        dp[0] = True                 # base case: empty prefix

        for i in range(1, n + 1):
            for j in range(i):
                if dp[j] and s[j:i] in word_set:
                    dp[i] = True
                    break

        return dp[n]
sobj = Solution()
sobj.wordBreak(s = "leetcode", wordDict = ["leet","code"])
```

Why it’s tricky

Some sequences are invalid:

"0" alone is invalid.

"06" is invalid (leading zeros not allowed).

We need to consider all combinations, not just one.


Possible decodings:

"2 2 6" → "BBF"

"22 6" → "VF"

"2 26" → "BZ" ✅ But "26" is valid

"226" as three-digit? → "226" is invalid, only max two digits

```python
from typing import List

class Solution:
    def numDecodings(self, s: str) -> int:
        # Edge case: empty string or starts with '0'
        if not s or s[0] == '0':
            return 0
        
        n = len(s)

        prev2, prev1 = 1, 1

        for i in range(1, n):
            curr = 0 
            if s[i] != '0':
                curr += prev1
            #two digits decode (10-26)
            two_digit = int(s[i-1: i+1])
            if 10 <= two_digit <= 26:
                curr += prev2
            

            prev2, prev1 = prev1, curr
        return prev1

if __name__ == "__main__":
    sobj = Solution()
    
    test_cases = [
        "12",       # "AB", "L" -> 2
        "226",      # "BBF", "VF", "BZ" -> 3
        "11106",    # "AAJF", "KJF" -> 2
        "0",        # invalid -> 0
        "10",       # "J" -> 1
        "27",       # "BG" -> 1 (27 > 26 cannot be two-digit)
        "",         # empty -> 0
        "1111",     # "AAAA", "KAA", "AKA", "AAK", "KK" -> 5
    ]

    for s in test_cases:
        result = sobj.numDecodings(s)
        print(f"Input: '{s}' -> Number of Decodings: {result}")
```

# 27 Clone Graph
Given a reference of a node in a connected undirected graph.
Return a deep copy (clone) of the graph.

Each node in the graph contains a value (int) and a list (List[Node]) of its neighbors.



```python
# Definition for a Node.
class Node:
    def __init__(self, val = 0, neighbors = None):
        self.val = val
        self.neighbors = neighbors if neighbors is not None else []

from typing import Optional
class Solution:
    def cloneGraph(self, node: Optional['Node']) -> Optional['Node']:
        if not node:
            return None

        visited = {}

        def dfs(current):
            if current in visited:
                return visited[current]

            # Clone current node
            clone = Node(current.val)
            visited[current] = clone

            # Clone all neighbors
            for neighbor in current.neighbors:
                clone.neighbors.append(dfs(neighbor))

            return clone

        return dfs(node)
        
```

```python
from typing import Optional
from collections import deque

class Node:
    def __init__(self, val=0, neighbors=None):
        self.val = val
        self.neighbors = neighbors if neighbors is not None else []

class Solution:
    def cloneGraph(self, node: Optional['Node']) -> Optional['Node']:
        if not node:
            return None

        visited = {}

        def dfs(current):
            if current in visited:
                return visited[current]

            clone = Node(current.val)
            visited[current] = clone

            for neighbor in current.neighbors:
                clone.neighbors.append(dfs(neighbor))

            return clone

        return dfs(node)

def build_graph(adj_list):
    if not adj_list:
        return None

    nodes = {}
    for i in range(1, len(adj_list) + 1):
        nodes[i] = Node(i)

    for i, neighbors in enumerate(adj_list, start=1):
        nodes[i].neighbors = [nodes[n] for n in neighbors]

    return nodes[1]

def print_graph(node):
    visited = set()
    queue = deque([node])

    while queue:
        curr = queue.popleft()
        if curr in visited:
            continue

        visited.add(curr)
        print(f"Node {curr.val}: {[n.val for n in curr.neighbors]}")

        for n in curr.neighbors:
            if n not in visited:
                queue.append(n)

# ---------- DRIVER CODE ----------
adj_list = [
    [2, 4],
    [1, 3],
    [2, 4],
    [1, 3]
]

original = build_graph(adj_list)

solution = Solution()
cloned = solution.cloneGraph(original)

print("Original Graph:")
print_graph(original)

print("\nCloned Graph:")
print_graph(cloned)

print("\nDeep Copy Check:")
print(original is cloned)
print(original.neighbors[0] is cloned.neighbors[0])
```

## 41 has Cycles

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


class Solution(object):
    def hasCycle(self, head):
        slow = head
        fast = head

        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next

            if slow == fast:
                return True
        return False


# Helper function to build a linked list with a cycle
def build_linked_list(values, pos):
    if not values:
        return None

    nodes = [ListNode(v) for v in values]

    for i in range(len(nodes) - 1):
        nodes[i].next = nodes[i + 1]

    if pos != -1:
        nodes[-1].next = nodes[pos]

    return nodes[0]


# ✅ Correct input
sobj = Solution()
head = build_linked_list([3, 2, 0, -4], 1)
print(sobj.hasCycle(head))  # True
```

Output: 4

```python
from collections import defaultdict
def characterReplacement(s: str, k: int) -> int:
    count = defaultdict(int)
    max_count = 0 
    res = 0 
    left = 0 
     

    for right in range(len(s)):
        count[s[right]] += 1
        max_count = max(max_count, count[s[right]])

        if (right -left +1) - max_count > k:
            count[s[left]] -=1
            left +=1
        res = max(res, right - left + 1)
    return res

print(characterReplacement("ABAB", 2))
```

### 58 647. Palindromic Substrings

```python
class Solution:
    def countSubstrings(self, s: str) -> int:
        n = len(s)
        count = 0 

        for center in range(2 * n -1):
            left = center // 2
            right  = left + center % 2

            while left >= 0 and right < n and s[left] == s[right]:
                count += 1
                left -= 1
                right += 1
        
        return count 
```

#### 647 Palindromic Substrings

```python
from typing import List
class Solution:
    def longestPalindrome(self, s: str) -> str:
        if not s:
            return ""

        start = 0
        max_len = 1

        def expand(left, right):
            nonlocal start, max_len
            while left >= 0 and right < len(s) and s[left] == s[right]:
                current_len = right - left + 1
                if current_len > max_len:
                    max_len = current_len
                    start = left
                left -=1
                right +=1
                

        for i in range(len(s)):
            expand(i, i)       # Odd length palindrome
            expand(i, i + 1)   # Even length palindrome

        return s[start:start + max_len]



sobj = Solution()
# nums = [100,4,200,1,3,2]
st = "babad"
sobj.longestPalindrome(st)
```
