```python
from typing import List
class Solution:
    def longestConsecutive(self, nums: List[int]) -> int:
        sor = sorted(nums)
        count = 1
        longest = 1
        print(sor)
        for i in range(1, len(sor)):
            if sor[i] == sor[i -1] + 1:
                count+=1
            
            elif sor[i] == sor[i -1]:
                continue

            else: 
                count = 1
            longest = max(longest, count)

        # print(longest)
        return longest



sobj = Solution()
nums = [100,4,200,1,3,2]
print(sobj.longestConsecutive(nums))
```
