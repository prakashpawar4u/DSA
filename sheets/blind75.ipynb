### 128. Longest Consecutive Sequence

```python
from typing import List
class Solution:
    def longestConsecutive(self, nums: List[int]) -> int:
        sor = sorted(nums)
        count = 1
        longest = 1
        print(sor)
        for i in range(1, len(sor)):
            if sor[i] == sor[i -1] + 1:
                count+=1
            
            elif sor[i] == sor[i -1]:
                continue

            else: 
                count = 1
            longest = max(longest, count)

        # print(longest)
        return longest



sobj = Solution()
nums = [100,4,200,1,3,2]
print(sobj.longestConsecutive(nums))
```

#### 1. Two Sum

```python
class Solution(object):
    def twoSum(self, nums, target):
        """
        :type nums: List[int]
        :type target: int
        :rtype: List[int]
        """
        hashMap = {}
        for i, n in enumerate(nums):
            diff = target - n
            if diff in hashMap:
                return(hashMap[diff], i)
            hashMap[n] = i
        return
        
```

#### 3. Longest Substring Without Repeating Characters

```python
class Solution:
    def lengthOfLongestSubstring(self, s: str) -> int:
        char_index = {}
        start = max_len = 0

        for end in range(len(s)):
            if s[end] in char_index and char_index[s[end]] >= start:
                start = char_index[s[end]] + 1
            char_index[s[end]] = end
            max_len = max(max_len, end - start + 1)
        return max_len

s = "abcabcbb"
sobj = Solution()
sobj.lengthOfLongestSubstring(s)
```

### 647. Palindromic Substrings

```python
class Solution:
    def countSubstrings(self, s: str) -> int:
        n = len(s)
        count = 0 

        for center in range(2 * n -1):
            left = center // 2
            right  = left + center % 2

            while left >= 0 and right < n and s[left] == s[right]:
                count += 1
                left -= 1
                right += 1
        
        return count 
```

### 133 Clone Graph

```python
from typing import Optional
from collections import deque

class Node:
    def __init__(self, val=0, neighbors=None):
        self.val = val
        self.neighbors = neighbors if neighbors is not None else []

class Solution:
    def cloneGraph(self, node: Optional['Node']) -> Optional['Node']:
        if not node:
            return None

        visited = {}

        def dfs(current):
            if current in visited:
                return visited[current]

            clone = Node(current.val)
            visited[current] = clone

            for neighbor in current.neighbors:
                clone.neighbors.append(dfs(neighbor))

            return clone

        return dfs(node)

def build_graph(adj_list):
    if not adj_list:
        return None

    nodes = {}
    for i in range(1, len(adj_list) + 1):
        nodes[i] = Node(i)

    for i, neighbors in enumerate(adj_list, start=1):
        nodes[i].neighbors = [nodes[n] for n in neighbors]

    return nodes[1]

def print_graph(node):
    visited = set()
    queue = deque([node])

    while queue:
        curr = queue.popleft()
        if curr in visited:
            continue

        visited.add(curr)
        print(f"Node {curr.val}: {[n.val for n in curr.neighbors]}")

        for n in curr.neighbors:
            if n not in visited:
                queue.append(n)

# ---------- DRIVER CODE ----------
adj_list = [
    [2, 4],
    [1, 3],
    [2, 4],
    [1, 3]
]

original = build_graph(adj_list)

solution = Solution()
cloned = solution.cloneGraph(original)

print("Original Graph:")
print_graph(original)

print("\nCloned Graph:")
print_graph(cloned)

print("\nDeep Copy Check:")
print(original is cloned)
print(original.neighbors[0] is cloned.neighbors[0])
```

#### 647 Palindromic Substrings

```python
from typing import List
class Solution:
    def longestPalindrome(self, s: str) -> str:
        if not s:
            return ""

        start = 0
        max_len = 1

        def expand(left, right):
            nonlocal start, max_len
            while left >= 0 and right < len(s) and s[left] == s[right]:
                current_len = right - left + 1
                if current_len > max_len:
                    max_len = current_len
                    start = left
                left -=1
                right +=1
                

        for i in range(len(s)):
            expand(i, i)       # Odd length palindrome
            expand(i, i + 1)   # Even length palindrome

        return s[start:start + max_len]



sobj = Solution()
# nums = [100,4,200,1,3,2]
st = "babad"
sobj.longestPalindrome(st)
```

### 11. Container With Most Water

```python
class Solution:
    def maxArea(self, heights: List[int]) -> int:
        res = 0
        l, r = 0, len(heights)-1

        while l < r:
            area = (r-l) * min(heights[l], heights[r])
            res = max(res, area)

            if heights[l] < heights[r]:
                l +=1
            else:
                r -=1
        return res
sobj = Solution()
height = [1,8,6,2,5,4,8,3,7]

sobj.maxArea(height)
```
